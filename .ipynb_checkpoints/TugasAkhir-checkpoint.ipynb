```python
# ============================================================
# IMPORT LIBRARY
# ============================================================
import pandas as pd
import numpy as np
import json
import re

import matplotlib.pyplot as plt
import seaborn as sns

from collections import Counter
from itertools import combinations

from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory

from sklearn.preprocessing import MultiLabelBinarizer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    multilabel_confusion_matrix
)

from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import LinearSVC
from sklearn.linear_model import LogisticRegression

from sklearn.pipeline import Pipeline, FeatureUnion

from iterstrat.ml_stratifiers import MultilabelStratifiedKFold

from skmultilearn.problem_transform import BinaryRelevance
```

```python
# ============================================================
# 1. LOAD DATASET
# ============================================================
data = []

with open('skripsi1_220711683_v2.jsonl', 'r', encoding='utf-8') as f:
    for line in f:
        data.append(json.loads(line))

df = pd.DataFrame(data)

print("===================================================")
print("DATASET")
print("===================================================")
print("Shape data :", df.shape)
print(df.head())
```

```python

# ============================================================
# 2. EXPLORATORY DATA ANALYSIS (EDA)
# ============================================================

# ------------------------------------------------------------
# 2.1 Distribusi Label
# ------------------------------------------------------------
all_labels = []

for labels in df['accept']:
    all_labels.extend(labels)

label_count = Counter(all_labels)

label_df = pd.DataFrame(
    label_count.items(),
    columns=['Label', 'Jumlah']
).sort_values(by='Jumlah', ascending=False)

print("\n===================================================")
print("DISTRIBUSI LABEL")
print("===================================================")
print(label_df)

# Visualisasi distribusi label
plt.figure(figsize=(10,5))
sns.barplot(
    x='Label',
    y='Jumlah',
    data=label_df
)

plt.xticks(rotation=45)
plt.title("Distribusi Label")
plt.xlabel("Label")
plt.ylabel("Jumlah")
plt.show()
```

```python
# ------------------------------------------------------------
# 2.2 Jumlah Label per Review
# ------------------------------------------------------------
df['jumlah_label'] = df['accept'].apply(len)

print("\n===================================================")
print("JUMLAH LABEL PER REVIEW")
print("===================================================")
print("Rata-rata label :", df['jumlah_label'].mean())

plt.figure(figsize=(8,5))
sns.histplot(df['jumlah_label'], bins=10)

plt.title("Distribusi Jumlah Label per Review")
plt.xlabel("Jumlah Label")
plt.ylabel("Jumlah Review")
plt.show()
```

```python
# ------------------------------------------------------------
# 2.3 Analisis Kombinasi Label
# ------------------------------------------------------------
pair_counter = Counter()
triple_counter = Counter()

for labels in df['accept']:

    labels = sorted(labels)

    # kombinasi 2 label
    pair_counter.update(combinations(labels, 2))

    # kombinasi 3 label
    triple_counter.update(combinations(labels, 3))

# ===== Pair =====
pair_df = pd.DataFrame(
    pair_counter.items(),
    columns=['Pair', 'Jumlah']
).sort_values(by='Jumlah', ascending=False)

pair_df[['Label_1', 'Label_2']] = pd.DataFrame(
    pair_df['Pair'].tolist(),
    index=pair_df.index
)

# ===== Triple =====
triple_df = pd.DataFrame(
    triple_counter.items(),
    columns=['Triple', 'Jumlah']
).sort_values(by='Jumlah', ascending=False)

triple_df[['Label_1', 'Label_2', 'Label_3']] = pd.DataFrame(
    triple_df['Triple'].tolist(),
    index=triple_df.index
)

print("\n===================================================")
print("TOP PAIR LABEL")
print("===================================================")
print(pair_df[['Label_1', 'Label_2', 'Jumlah']].head(5))

print("\n===================================================")
print("TOP TRIPLE LABEL")
print("===================================================")
print(triple_df[['Label_1', 'Label_2', 'Label_3', 'Jumlah']].head(5))
```

```python
# ------------------------------------------------------------
# 2.4 Analisis Panjang Review
# ------------------------------------------------------------
df['panjang_review'] = df['text'].apply(
    lambda x: len(x.split())
)

print("\n===================================================")
print("PANJANG REVIEW")
print("===================================================")
print("Rata-rata panjang review :", df['panjang_review'].mean())

plt.figure(figsize=(10,5))
sns.histplot(df['panjang_review'], bins=50)

plt.title("Distribusi Panjang Review")
plt.xlabel("Jumlah Kata")
plt.ylabel("Jumlah Data")
plt.show()

# Kategori panjang review
short_review = df[df['panjang_review'] < 10]
medium_review = df[
    (df['panjang_review'] >= 10) &
    (df['panjang_review'] <= 30)
]
long_review = df[df['panjang_review'] > 30]

print("\n===================================================")
print("KATEGORI REVIEW")
print("===================================================")
print("Review pendek :", len(short_review))
print("Review sedang :", len(medium_review))
print("Review panjang:", len(long_review))
```

```python
plt.figure(figsize=(10,5))
sns.barplot(x='Label', y='Jumlah', data=label_df)
plt.xticks(rotation=45)
plt.title("Distribusi Label")
plt.show()
```

```python
# ============================================================
# 3. DATA CLEANING
# ============================================================

# ------------------------------------------------------------
# 3.1 Menghapus Label diluar_konteks
# ------------------------------------------------------------
df = df[
    ~df['accept'].apply(
        lambda x: 'diluar_konteks' in x
    )
]

print("\n===================================================")
print("SETELAH HAPUS LABEL 'diluar_konteks'")
print("===================================================")
print("Jumlah data :", len(df))

label_count = Counter([label for labels in df['accept'] for label in labels])

print(label_count)
```

```python
# ============================================================
# 4. TEXT PREPROCESSING
# ============================================================

# ------------------------------------------------------------
# 4.1 Stopword Removal
# ------------------------------------------------------------

stop_factory = StopWordRemoverFactory()
stopwords = stop_factory.get_stop_words()

# lihat jumlah stopwords
print("Jumlah stopwords:", len(stopwords))

# lihat sebagian stopwords
print("Contoh stopwords:", stopwords)
```

```python
# daftar kata negasi yang ingin dipertahankan
negasi = ['tidak', 'bukan', 'nggak', 'tanpa', 'kecuali', 'belum']

custom_stopwords = set(stopwords)

# hapus kata negasi dari stopwords
for kata in negasi:
    custom_stopwords.discard(kata)

# ------------------------------------------------------------
# 4.2 Fungsi Preprocessing
# ------------------------------------------------------------
def preprocess(text):
    tokens = re.findall(r'\b\w+\b', text.lower())  # tokenisasi TANPA split()
    return ' '.join(w for w in tokens if w not in custom_stopwords)

# ------------------------------------------------------------
# 4.3 Terapkan Preprocessing
# ------------------------------------------------------------
df_final = df.copy()

df_final['clean_text'] = df_final['text'].apply(preprocess)

# tampilkan
for i in range(5):
    print("Original :", df_final['text'].iloc[i])
    print("Clean    :", df_final['clean_text'].iloc[i])
    print("-----")
```

```python
# ============================================================
# 5. TRANSFORM LABEL MULTILABEL
# ============================================================
mlb = MultiLabelBinarizer()

y = mlb.fit_transform(df_final['accept'])

print("\n===================================================")
print("MULTILABEL BINARIZER")
print("===================================================")
print("Label classes:", mlb.classes_)
print("Shape y:", y.shape)
```

```python
# ============================================================
# 6. SPLIT DATA
# ============================================================
X = df_final['clean_text'].values
y = y  # hasil MultiLabelBinarizer

mskf = MultilabelStratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# ambil 1 fold untuk test
for train_idx, test_idx in mskf.split(X, y):
    X_train = X[train_idx]
    X_test = X[test_idx]

    y_train = y[train_idx]
    y_test = y[test_idx]
    
    break  # ambil fold pertama saja
```

```python
# ============================================================
# 7. TF-IDF PIPELINE
# ============================================================

tfidf_pipeline = FeatureUnion([

    # ========================================================
    # WORD TF-IDF
    # ========================================================
    ('word_tfidf', TfidfVectorizer(
        max_features=5000,
        ngram_range=(1,2),
        min_df=2,
        max_df=0.9,
        sublinear_tf=True
    )),

    # ========================================================
    # CHARACTER TF-IDF
    # ========================================================
    ('char_tfidf', TfidfVectorizer(
        max_features=3000,
        analyzer='char_wb',
        ngram_range=(3,4),
        min_df=2,
        sublinear_tf=True
    ))

])
```

```python
# ============================================================
# 8. MODELING
# ============================================================

results = {}

# ============================================================
# 8.1 NAIVE BAYES
# ============================================================
print("\n===================================================")
print("TRAINING NAIVE BAYES")
print("===================================================")

nb_pipeline = Pipeline([

    ('tfidf', tfidf_pipeline),

    ('clf', BinaryRelevance(
        MultinomialNB()
    ))

])

param_nb = {

    'clf__classifier__alpha': [0.1, 0.5, 1.0, 2.0]

}

grid_nb = GridSearchCV(
    estimator=nb_pipeline,
    param_grid=param_nb,
    scoring='f1_micro',
    cv=5,
    n_jobs=-1,
    verbose=1
)

grid_nb.fit(X_train, y_train)

results['Naive Bayes'] = grid_nb.best_score_

print("Best Params :", grid_nb.best_params_)
print("Best F1     :", grid_nb.best_score_)

# ============================================================
# 8.2 SUPPORT VECTOR MACHINE
# ============================================================
print("\n===================================================")
print("TRAINING SVM")
print("===================================================")

svm_pipeline = Pipeline([

    ('tfidf', tfidf_pipeline),

    ('clf', BinaryRelevance(
        LinearSVC()
    ))

])

param_svm = {

    'clf__classifier__C': [0.5, 1.0, 2.0],

    'clf__classifier__loss': [
        'hinge',
        'squared_hinge'
    ],

    'clf__classifier__class_weight': [
        'balanced'
    ]

}

grid_svm = GridSearchCV(
    estimator=svm_pipeline,
    param_grid=param_svm,
    scoring='f1_micro',
    cv=5,
    n_jobs=-1,
    verbose=1
)

grid_svm.fit(X_train, y_train)

results['SVM'] = grid_svm.best_score_

print("Best Params :", grid_svm.best_params_)
print("Best F1     :", grid_svm.best_score_)

# ============================================================
# 8.3 LOGISTIC REGRESSION
# ============================================================
print("\n===================================================")
print("TRAINING LOGISTIC REGRESSION")
print("===================================================")

lr_pipeline = Pipeline([

    ('tfidf', tfidf_pipeline),

    ('clf', BinaryRelevance(
        LogisticRegression(
            max_iter=2000
        )
    ))

])

param_lr = {

    'clf__classifier__C': [2.0, 5.0],

    'clf__classifier__solver': [
        'lbfgs'
    ],

    'clf__classifier__class_weight': [
        'balanced'
    ]

}

grid_lr = GridSearchCV(
    estimator=lr_pipeline,
    param_grid=param_lr,
    scoring='f1_micro',
    cv=5,
    n_jobs=-1,
    verbose=1
)

grid_lr.fit(X_train, y_train)

results['Logistic Regression'] = grid_lr.best_score_

print("Best Params :", grid_lr.best_params_)
print("Best F1     :", grid_lr.best_score_)
```

```python
# ============================================================
# 9. PERBANDINGAN HASIL MODEL
# ============================================================
print("\n===================================================")
print("HASIL SEMUA MODEL")
print("===================================================")

for name, score in results.items():
    print(f"{name}: {score:.4f}")
```

```python
# ============================================================
# 10. EVALUASI MODEL TERBAIK
# ============================================================

# gunakan model terbaik
best_model = grid_lr.best_estimator_
# training ulang
best_model.fit(X_train, y_train)
# prediksi data test
y_pred_test = best_model.predict(X_test)

print("\n=== HASIL TEST ===")

# ------------------------------------------------------------
# 10.1 F1 Score
# ------------------------------------------------------------
# ============================================================
# EVALUASI METRIK
# ============================================================

micro_precision = precision_score(
    y_test,
    y_pred_test,
    average='micro',
    zero_division=0
)

micro_recall = recall_score(
    y_test,
    y_pred_test,
    average='micro',
    zero_division=0
)

macro_precision = precision_score(
    y_test,
    y_pred_test,
    average='macro',
    zero_division=0
)

macro_recall = recall_score(
    y_test,
    y_pred_test,
    average='macro',
    zero_division=0
)

micro_f1 = f1_score(
    y_test,
    y_pred_test,
    average='micro'
)

macro_f1 = f1_score(
    y_test,
    y_pred_test,
    average='macro'
)

print("\n===================================================")
print("HASIL EVALUASI")
print("===================================================")

print("Micro Precision :", micro_precision)
print("Micro Recall    :", micro_recall)
print("Micro F1-Score  :", micro_f1)

print()

print("Macro Precision :", macro_precision)
print("Macro Recall    :", macro_recall)
print("Macro F1-Score  :", macro_f1)

# ------------------------------------------------------------
# 10.2 Classification Report
# ------------------------------------------------------------
print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred_test,
    target_names=mlb.classes_,
    zero_division=0
))
```

```python
# ============================================================
# 11. CONFUSION MATRIX
# ============================================================

cm = multilabel_confusion_matrix(y_test, y_pred_test)

labels = mlb.classes_

for i in range(len(labels)):
    plt.figure(figsize=(4,3))
    sns.heatmap(cm[i], annot=True, fmt='d', cmap='Blues')
    plt.title(f'Confusion Matrix: {labels[i]}')
    plt.xlabel('Predicted')
    plt.ylabel('Actual')
    plt.show()
```

```python
# ============================================================
# FUNGSI PREDIKSI
# ============================================================
def predict_text(text):

    # preprocessing
    clean = preprocess(text)

    # prediksi langsung menggunakan pipeline
    pred = best_model.predict([clean])

    # ubah hasil multilabel menjadi nama label
    labels = mlb.inverse_transform(pred)

    return labels[0]

# ============================================================
# CONTOH PREDIKSI
# ============================================================
text_input = "csnya ga responsif"

hasil = predict_text(text_input)

print("\n===================================================")
print("HASIL PREDIKSI")
print("===================================================")

print("Input     :", text_input)
print("Prediksi  :", hasil)
```
